```python
import pandas as pd
import os

# Define the file path to the building stock CSV file
file_path = os.path.join("input", "stock", "buildingstock_sdes2018_update_hpdiff_ac_inseecalibrated_zcl.csv")

# Read the building stock table
df = pd.read_csv(file_path)

# Check if the column already exists to avoid duplicate addition
if 'roof_albedo' not in df.columns:
    # Add roof_albedo to the last column, defaulting to 0.2 (20% reflectance, dark roof)
    df['roof_albedo'] = 0.2
    
    # Keep the original index, don't retain the automatically generated index column
    df.to_csv(file_path, index=False)
    print(f"Successfully added 'roof_albedo' column with default value 0.2. Total rows processed: {len(df)}.")
else:
    print("'roof_albedo' column already exists, no need to add it again.")
```
